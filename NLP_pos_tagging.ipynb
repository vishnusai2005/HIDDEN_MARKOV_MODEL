```python
train_data = [
    [("Mary", "N"), ("Jane", "N"), ("can", "M"), ("see", "V"), ("Will", "N")],
    [("Spot", "N"), ("will", "M"), ("see", "V"), ("Mary", "N")],
    [("Will", "M"), ("Jane", "N"), ("spot", "V"), ("Mary", "N")],
    [("Mary", "N"), ("will", "M"), ("pat", "V"), ("Spot", "N")]
]
```

```python
from collections import defaultdict 
emission_counts = defaultdict(lambda: defaultdict(int)) 
transition_counts = defaultdict(lambda: defaultdict(int))
start_counts = defaultdict(int)
tag_counts = defaultdict(int) 
for sent in train_data:
    previous_tag = "<S>"
    start_counts[sent[0][1]] += 1
    for word, tag in sent:
        emission_counts[tag][word.lower()] += 1 
        tag_counts[tag] += 1
        transition_counts[previous_tag][tag] += 1 
        previous_tag = tag 
    transition_counts[previous_tag]["<E>"] += 1


#print(emission_counts)
#print(start_counts)
#print(tag_counts)
print(previous_tag)
#print(transition_counts) 


```

```python
#Normalize Counts 
def normalize(d):
    total = sum(d.values())
    return {k: v / total for k, v in d.items()}

start_prob = normalize(start_counts)
transition_prob = {k: normalize(v) for k, v in transition_counts.items()}
emission_prob = {k: normalize(v)for k, v in emission_counts.items()}
print("Start Probabilities:", start_prob)
print("Transition Probabilities:", transition_prob)
print("Emission Probabilities:", emission_prob)
```

```python
#tagging a sentence - calculates sequence probability 
sentence = ['Will', "can", "spot", "Mary"]
tags = ["N", "M", "V", "N"]
p = start_prob.get(tags[0], 0)
if p == 0:
    print("Zero start probability - impossible sequence")
else:
    for i , word in enumerate(sentence):
        tag = tags[i]
        emission_p = emission_prob[tag].get(word.lower(), 0)
        if emission_p == 0:
            print(f"Emission probability zero for {word}/{tag} - impossible sequence")
            p = 0 
            break
        p *= emission_p 
        if i < len(tags) - 1:
            trans_p = transition_prob[tags[i]].get(tags[i+1], 0) 
            if trans_p == 0:
                print(f"Transition probability zero for {tags[i]} -> {tags[i+1]} - impossible sequence")
                p = 0 
                break
            p *= trans_p 
    if p > 0:
        p *= transition_prob[tags[-1]].get("<E>",0)
print(f"Probability of sequence {tags} : {p}")
```
